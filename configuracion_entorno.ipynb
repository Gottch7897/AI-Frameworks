```python
from pathlib import Path

import matplotlib
import sklearn
import tensorflow as tf
import torch
import torchvision
from icrawler.builtin import BingImageCrawler, GoogleImageCrawler
from imagededup.methods import PHash
from PIL import Image, ImageOps

PROJECT_ROOT = Path('.').resolve()
DATA_ROOT = PROJECT_ROOT / 'data'
ARTIFACTS_ROOT = PROJECT_ROOT / 'artifacts'
for folder in (DATA_ROOT, ARTIFACTS_ROOT):
    folder.mkdir(parents=True, exist_ok=True)

print('Proyecto:', PROJECT_ROOT)
print('Data root:', DATA_ROOT)
print('Artifacts root:', ARTIFACTS_ROOT)
print('Torch:', torch.__version__)
print('Torchvision:', torchvision.__version__)
print('TensorFlow:', tf.__version__)
print('Scikit-learn:', sklearn.__version__)
print('Matplotlib:', matplotlib.__version__)
print('GPU disponible en PyTorch:', torch.cuda.is_available())
print('GPUs visibles en TensorFlow:', tf.config.list_physical_devices('GPU'))
```

```python
# Ajustes utiles despues de instalar las dependencias
# Si hay GPU, TensorFlow usara memoria incremental en vez de reservarla toda al inicio
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    for gpu in gpus:
        tf.config.experimental.set_memory_growth(gpu, True)
    print('TensorFlow configurado para crecimiento progresivo de memoria GPU.')
else:
    print('No se detecto GPU. El flujo seguira en CPU.')

# Semilla basica para reproducibilidad
SEED = 42
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
tf.keras.utils.set_random_seed(SEED)
print('Semilla aplicada:', SEED)
```
