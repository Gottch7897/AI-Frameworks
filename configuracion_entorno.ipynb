```python
from pathlib import Path

import matplotlib
import sklearn
import torch
from icrawler.builtin import BingImageCrawler, GoogleImageCrawler
from imagededup.methods import PHash
from PIL import Image, ImageOps

PROJECT_ROOT = Path('.').resolve()
DATA_ROOT = PROJECT_ROOT / 'data'
ARTIFACTS_ROOT = PROJECT_ROOT / 'artifacts'
for folder in (DATA_ROOT, ARTIFACTS_ROOT):
    folder.mkdir(parents=True, exist_ok=True)

print('Proyecto:', PROJECT_ROOT)
print('Data root:', DATA_ROOT)
print('Artifacts root:', ARTIFACTS_ROOT)
print('Torch:', torch.__version__)
print('Scikit-learn:', sklearn.__version__)
print('Matplotlib:', matplotlib.__version__)
print('GPU disponible en PyTorch:', torch.cuda.is_available())
```

```python
# Ajustes utiles para el flujo de PyTorch

# Semilla basica para reproducibilidad
SEED = 42
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
print('Semilla aplicada:', SEED)
```
